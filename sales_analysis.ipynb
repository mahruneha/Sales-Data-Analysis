This beginner-level project analyzes a sample sales dataset to understand revenue trends, product performance, category performance, and regional sales.

## 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
```

## 2. Load the Dataset

```python
df = pd.read_csv('sales_data.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.head()
```

## 3. Basic Information

```python
print('Rows:', len(df))
print('Columns:', len(df.columns))
print('\nMissing values:')
print(df.isnull().sum())
print('\nData types:')
print(df.dtypes)
```

## 4. Key Performance Indicators

```python
total_revenue = df['Revenue'].sum()
total_units = df['Quantity'].sum()
average_order_value = df['Revenue'].mean()

print(f'Total Revenue: ${total_revenue:,.2f}')
print(f'Total Units Sold: {total_units:,}')
print(f'Average Transaction Revenue: ${average_order_value:,.2f}')
```

## 5. Monthly Revenue

```python
df['Month'] = df['Date'].dt.to_period('M').astype(str)
monthly_sales = df.groupby('Month', as_index=False)['Revenue'].sum()

plt.figure(figsize=(10,5))
plt.plot(monthly_sales['Month'], monthly_sales['Revenue'], marker='o')
plt.title('Monthly Revenue')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6. Product Performance

```python
product_sales = df.groupby('Product', as_index=False)['Revenue'].sum().sort_values('Revenue', ascending=False)
display(product_sales)

plt.figure(figsize=(9,5))
plt.bar(product_sales['Product'], product_sales['Revenue'])
plt.title('Revenue by Product')
plt.xlabel('Product')
plt.ylabel('Revenue')
plt.xticks(rotation=35, ha='right')
plt.tight_layout()
plt.show()
```

## 7. Category Performance

```python
category_sales = df.groupby('Category', as_index=False)['Revenue'].sum().sort_values('Revenue', ascending=False)
display(category_sales)

plt.figure(figsize=(8,5))
plt.bar(category_sales['Category'], category_sales['Revenue'])
plt.title('Revenue by Category')
plt.xlabel('Category')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

## 8. Regional Performance

```python
region_sales = df.groupby('Region', as_index=False)['Revenue'].sum().sort_values('Revenue', ascending=False)
display(region_sales)

plt.figure(figsize=(8,5))
plt.bar(region_sales['Region'], region_sales['Revenue'])
plt.title('Revenue by Region')
plt.xlabel('Region')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

## 9. Key Findings

- Compare monthly revenue to identify stronger and weaker periods.
- Identify the products that generate the most revenue.
- Compare categories to understand which business area contributes most to revenue.
- Compare regions to understand geographic sales performance.
